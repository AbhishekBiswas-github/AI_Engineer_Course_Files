````python
import json
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    model="groq:llama-3.3-70b-versatile"
)

with open("output.txt", "r", encoding="utf-8") as f:
    text = f.read()

response = model.invoke([
    {
        "role": "system",
        "content": "You are an intelligent agent. You analyze text and return structured JSON only."
    },
    {
        "role": "user",
        "content": f"""Analyze the following text and return ONLY valid JSON:

{text}

Format:
{{
    "Document Number": "",
    "subject": "",
    "date": "",
    "person_signature": "",
    "subject_line": "",
    "summary": ""
}}
"""
    }
])

# ✅ Step 1: Extract model output
output_text = response.content.strip()

# ✅ Step 2: Convert to dictionary safely
try:
    data = json.loads(output_text)
except json.JSONDecodeError:
    print("⚠️ Invalid JSON returned. Attempting cleanup...")

    # Optional cleanup (basic fix for common issues)
    cleaned_text = output_text.replace("```json", "").replace("```", "").strip()

    try:
        data = json.loads(cleaned_text)
    except json.JSONDecodeError:
        print("❌ Failed to parse JSON even after cleanup.")
        data = None

# ✅ Step 3: Use structured dict safely
if data:
    print("✅ Parsed Dictionary:")
    print(data)

    print("\n📌 Extracted Fields:")
    print("Document Number:", data.get("Document Number"))
    print("Subject:", data.get("subject"))
    print("Date:", data.get("date"))
    print("Signature:", data.get("person_signature"))
    print("Subject Line:", data.get("subject_line"))
    print("Summary:", data.get("summary"))
````
