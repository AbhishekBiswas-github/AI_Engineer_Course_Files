```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    model = "groq:llama-3.3-70b-versatile"
)

model
```

```python
from typing_extensions import TypedDict, Annotated


class Movie(TypedDict):
    """A movie with details"""
    title: Annotated[str, ..., "Title of the movie"]
    rating: Annotated[float, ..., "Rating of the movie"]
    director: Annotated[str, ..., "Director of the movie"]
    year: Annotated[int, ..., "Year on which the movie released"]

model_with_structure = model.with_structured_output(Movie)
response = model_with_structure.invoke("Provide me the information of the movie Inception")
response

```

```python
# Nested strcuture
class Actor(TypedDict):
    name:str
    role:str

class MovieDetails(TypedDict):
    Title:str
    Year:int
    Director:str
    Cast: Annotated[list[Actor], ..., "List of cast, minimum 10"]
    Genres: list[str]
    Budget: Annotated[float, ..., "Budget of the movie in millions USD."]

model_with_structure = model.with_structured_output(MovieDetails)
response = model_with_structure.invoke("Provide me the information of the movie Avengers Infinity Stones")
response
```
